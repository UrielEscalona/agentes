```python
import numpy as np
import random
import matplotlib.pyplot as plt
import time
from IPython.display import clear_output

# Configuración para matplotlib en Colab
%matplotlib inline
```

```python
# Tamaño del tablero
GRID_SIZE = 9
CELL_SIZE = 50

# Sudoku predefinido (0 representa casillas vacías)
sudoku_puzzle = [
    [5, 3, 0, 0, 7, 0, 0, 0, 0],
    [6, 0, 0, 1, 9, 5, 0, 0, 0],
    [0, 9, 8, 0, 0, 0, 0, 6, 0],
    [8, 0, 0, 0, 6, 0, 0, 0, 3],
    [4, 0, 0, 8, 0, 3, 0, 0, 1],
    [7, 0, 0, 0, 2, 0, 0, 0, 6],
    [0, 6, 0, 0, 0, 0, 2, 8, 0],
    [0, 0, 0, 4, 1, 9, 0, 0, 5],
    [0, 0, 0, 0, 8, 0, 0, 7, 9]
]
```

```python
# Posiciones de los números fijos
fixed_positions = {(r, c) for r in range(GRID_SIZE) for c in range(GRID_SIZE) if sudoku_puzzle[r][c] != 0}

def draw_grid(grid, generation, fitness_score):
    """Dibuja el Sudoku usando Matplotlib"""
    clear_output(wait=True)
    plt.figure(figsize=(8, 8))
    plt.title(f"Generación: {generation} - Fitness: {fitness_score}", fontsize=14)
    #plt.imshow(np.ones((GRID_SIZE, GRID_SIZE)) * 0.1, cmap="gray", extent=[0, GRID_SIZE, 0, GRID_SIZE])
    plt.gca().set_facecolor((0.8, 0.8, 0.8))  # RGB: 0.8 es un gris claro

    # Dibujar cuadrícula
    for r in range(GRID_SIZE + 1):
        lw = 2 if r % 3 == 0 else 0.5
        plt.plot([0, GRID_SIZE], [r, r], 'k', linewidth=lw)
        plt.plot([r, r], [0, GRID_SIZE], 'k', linewidth=lw)

    # Añadir números
    for r in range(GRID_SIZE):
        for c in range(GRID_SIZE):
            if grid[r][c] != 0:
                color = 'darkred' if (r, c) in fixed_positions else 'darkblue'
                plt.text(c + 0.5, GRID_SIZE - r - 0.5, str(grid[r][c]),
                        ha='center', va='center', fontsize=14, color=color, fontweight='bold')

    plt.xticks([])
    plt.yticks([])
    plt.show()
    plt.close()  # Cerrar figura para liberar memoria

def generate_population(size):
    """Genera población inicial respetando números fijos"""
    return [np.array([[sudoku_puzzle[r][c] if (r,c) in fixed_positions
                      else random.randint(1, 9) for c in range(9)] for r in range(9)])
            for _ in range(size)]

def fitness(grid):
    """Calcula fitness basado en números únicos por filas, columnas y grid 3x3 """
    row_unique = sum(len(set(row)) for row in grid)
    col_unique = sum(len(set(col)) for col in grid.T)

    grid_size = 3
    grid_3x3 = [
        grid[i*grid_size:(i + 1)*grid_size, 
            j*grid_size:(j + 1)*grid_size] 
            for j in range(3) 
            for i in range(3)
    ]
    
    grid_3x3_uniques = sum(len(np.unique(arr)) for arr in grid_3x3)
    
    return row_unique + col_unique + grid_3x3_uniques


def crossover(p1, p2):
    """Cruzamiento por filas con probabilidad 50%"""
    mask = np.random.rand(9) < 0.5
    type_cross = np.random.randint(3)

    if type_cross == 0: #  Cruzamiendo por fila
        child = np.where(mask[:, None], p1, p2)
    elif type_cross == 1: # Cruzamiendo por columna
        child =  np.where(mask[None, :], p1, p2)
    else: # Cruzamiendo  por bloque
        mask_3x3 = np.repeat(
            np.repeat(mask.reshape(3, 3), 3, axis=0),
        3, axis=1)
        child = np.where(mask_3x3, p1, p2)
    
    return child

def mutate(grid, mutation_rate=0.05):
    """Mutación respetando números fijos por intercambio de filas y columnas"""

    mask = np.random.rand(9) < mutation_rate
    # Intercambiar filas
    for r in range(9):
        if mask[r]:
            cols = [c for c in range(9) if (r, c) not in fixed_positions]
            selected_cols = np.random.choice(cols, size=2, replace=False)
            aux = grid[r,selected_cols[0]]
            grid[r,selected_cols[0]] = grid[r,selected_cols[1]]
            grid[r,selected_cols[1]] = aux

    # Intercambiar columnas
    mask = np.random.rand(9) < mutation_rate
    for c in range(9):
        if mask[c]:
            rows = [r for r in range(9) if (r, c) not in fixed_positions]
            selected_rows = np.random.choice(rows, size=2, replace=False)
            aux = grid[selected_rows[0], r]
            grid[selected_rows[0], r] = grid[selected_rows[1], r]
            grid[selected_rows[1], r] = aux

    return grid

def mutate_original(grid, mutation_rate=0.05):
    """Mutación respetando números fijos"""
    
    np.array([[(r,c) in fixed_positions for c in range(9)] for r in range(9)])

    mutation_mask = np.random.rand(*grid.shape) < mutation_rate
    random_values = np.random.randint(1, 10, grid.shape)
    return np.where(mutation_mask & ~np.array([[ (r,c) in fixed_positions
                                               for c in range(9)] for r in range(9)]),
                   random_values, grid)

def genetic_algorithm(pop_size=100, max_generations=1000, mutation_rate=0.05):
    """Algoritmo genético principal"""
    population = generate_population(pop_size)
    best_fitness = 0

    for generation in range(max_generations):
        # Evaluar y ordenar población
        population.sort(key=lambda x: -fitness(x))
        best_solution = population[0]
        current_fitness = fitness(best_solution)

        # Mostrar progreso cada 10 generaciones
        if generation % 10 == 0:
            draw_grid(best_solution, generation, current_fitness)
            print(f"Generación: {generation:4d} | Fitness: {current_fitness:3d} | Máximo posible: 243")
            time.sleep(0.1)

        # Condición de término
        if current_fitness == 243:
            print("\n¡Sudoku resuelto!")
            draw_grid(best_solution, generation, current_fitness)
            return best_solution

        # Selección y reproducción
        parents = population[:pop_size//2]  # Seleccionar mejores 50%
        new_pop = [best_solution.copy()]

        while len(new_pop) < pop_size:
            p1, p2 = random.choices(parents, k=2)
            child = mutate(crossover(p1, p2), mutation_rate)
            new_pop.append(child)

        population = new_pop

    print("\nMáximo de generaciones alcanzado. Solución no encontrada.")
    draw_grid(best_solution, max_generations, best_fitness)
    return best_solution
```

```python
# Ejecutar el algoritmo con parámetros ajustados
genetic_algorithm(pop_size=3000, max_generations=300, mutation_rate=0.03)
```
